# Number 4: Largest Palindrome Product
A palindromic number reads the same both ways. The largest palindrome made from the product of two $2$-digit numbers is $9009=91\times99$ .

Find the largest palindrome made from the product of two $3$-digit numbers.

```python
import numpy as np

def isPalindrome(num):
    listDig = [int(d) for d in str(num)]
    rev = listDig[::-1]
    return (rev == listDig)

def biggestPalindrome(limLow, limHigh):
    palList = []
    factors = np.arange(limHigh, limLow-1, -1)
    for f1 in factors:
        for f2 in factors:
            if isPalindrome(f1*f2):
                palList.append(f1*f2)
    return max(palList)

biggestPalindrome(100, 999)
```

# Number 5: Smallest Multiple

$2520$ is the smallest number that can be divided by each of the numbers from $1$ to $10$ without any remainder.

What is the smallest positive number that is **evenly divisible** by all of the numbers from $1$ to $20$?

```python
import numpy as np 

def smallest_multiple(low, high):
    candidates = np.arange(2520, 1e9, 2520)
    listNums = np.array(np.arange(low, high+1, 1))
    sums = [(sum(objs%listNums) + (objs%2)) for objs in candidates]
    return candidates[sums.index(0)]

smallest_multiple(1, 20)
```

# Number 6: Sum Square Difference

The sum of the squares of the first ten natural numbers is,
$$ 1^2 + 2^2 + ... + 10^2 = 385.$$

The square of the sum of the first ten natural numbers is,
$$ (1 + 2 + ... + 10)^2 = 55^2 = 3025.$$

Hence the difference between the sum of the squares of the first ten natural numbers and the square of the sum is $3025-385=2640$.

Find the difference between the sum of the squares of the first one hundred natural numbers and the square of the sum.

```python
import numpy as np

def sum_square_difference(naturalNums):
    listNums = np.arange(1,naturalNums+1,1)
    sumSquares = sum(listNums*listNums)
    squareSum = sum(listNums)**2
    return (squareSum - sumSquares)

sum_square_difference(100)
```

# Number 7: 10 001st Prime
By listing the first six prime numbers: $2,3,5,7,11,$ and $13$, we can see that the $6$th prime is $13$.

What is the $10001$st prime number?

```python
import numpy as np
from math import prod

# function to check whether a number is prime but it is so slow
def isPrime(num):
    divisors = np.arange(2, num**.5, 1)
    removedNum = divisors[divisors != num]
    has_zero = 0 in [num%d for d in removedNum]
    if not has_zero:
        return True
    else:
        return False

# slowly climb up via counter reaching the 10 001st prime
order = 10001
counter = 0
number = 2
while counter < (order):
    if isPrime(number):
        counter += 1
        number += 1
    else:
        number += 1

# this is very primitive implementation
print(number - 1)
```

# Problem 9: Special Pythagorean Triplet

<p>A Pythagorean triplet is a set of three natural numbers, $a \lt b \lt c$, for which,
$$a^2 + b^2 = c^2.$$</p>
<p>For example, $3^2 + 4^2 = 9 + 16 = 25 = 5^2$.</p>
<p>There exists exactly one Pythagorean triplet for which $a + b + c = 1000$.<br>Find the product $abc$.</p>

```python
import numpy as np

sumTriplet = 1000

row = np.array(np.arange(1, sumTriplet, 1))

matrix = [row, row, row]

# given the sum value condition, identify the total number of possible combinations of a, b, and c
def combiCalculator(sumTriplet):
    dimTotal = 0
    for i in range(1, sumTriplet):
        dimTotal += sum(range(1,i))
    return dimTotal

dim = combiCalculator(sumTriplet)
newMat = np.zeros((dim, 5)) # number of rows represent the combinations and cols represent the 3 values and 2 slots for the conditions

i = 0
for a in range(1, sumTriplet+1):
    for b in range(a+1, sumTriplet+1):
        for c in range(b+1, sumTriplet+1):
            newMat[i][0] = a
            newMat[i][1] = b
            newMat[i][2] = c
            newMat[i][3] = (a**2 + b**2 == c**2)*1.0
            newMat[i][4] = (a+b+c == sumTriplet)*1.0
            i += 1
            
tripletInd = np.nonzero(np.prod(newMat, axis=1))
tripletInd = tripletInd[0][0]
print(fr'The Pythagorean triplet for which a+b+c={sumTriplet} are...')
print(fr'a={newMat[tripletInd][0]}, b={newMat[tripletInd][1]}, c={newMat[tripletInd][2]}.')
print(fr'Their product is abc={newMat[tripletInd][0]*newMat[tripletInd][1]*newMat[tripletInd][2]}')
```
