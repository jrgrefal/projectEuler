# Number 3

The prime factors of $13195$ are $5, 7, 13,$ and $29$.

What is the largest prime factor of the number $600851475143$?

```python
num = 13195
5*7*13*29
```

```python
def primeFactorsCalc(num, itermax):
    primeFactors = []
    for i in range(2, itermax):
        if num%i == 0:
            num /= i
            primeFactors.append(i)
    return primeFactors
```

```python
max(primeFactorsCalc(600851475143, itermax=1000000))
```
