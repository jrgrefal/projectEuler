# Problem 0

A number is a perfect square, or a square number, if it is the square of a positive integer.
For example, $25$ is a square number because $5^2=5\times5=25$; it is also an odd square.

The first 5 square numbers are: $1,4,9, 16,$ and $25$ and the sum of the odd squares is $1+9+25=35$.

Among the first **613 thousand** square numbers, what is the sum of all the odd squares?

```python
def sumFirstNOddSquares(N):
    ''' Among the first N square numbers, calculate the sum of all the odd squares.
    Input: how many squares to start off of from 1
    Output: sum of all first N odd squares'''
    
    listRoots = np.arange(1, N+1, 1) # create the list
    listSquares = listRoots**2 # get first N squares
    listSquares[listSquares%2 == 0] = 0 # make all even values to zero
    return np.sum(listSquares) # sum all nonzero squares

sumFirstNOddSquares(613e3)
```

# Number 4: Largest Palindrome Product
A palindromic number reads the same both ways. The largest palindrome made from the product of two $2$-digit numbers is $9009=91\times99$ .

Find the largest palindrome made from the product of two $3$-digit numbers.

```python
import numpy as np

def isPalindrome(num):
    listDig = [int(d) for d in str(num)]
    rev = listDig[::-1]
    return (rev == listDig)

def biggestPalindrome(limLow, limHigh):
    palList = []
    factors = np.arange(limHigh, limLow-1, -1)
    for f1 in factors:
        for f2 in factors:
            if isPalindrome(f1*f2):
                palList.append(f1*f2)
    return max(palList)

biggestPalindrome(100, 999)
```

# Number 5: Smallest Multiple

$2520$ is the smallest number that can be divided by each of the numbers from $1$ to $10$ without any remainder.

What is the smallest positive number that is **evenly divisible** by all of the numbers from $1$ to $20$?

```python
import numpy as np 

def smallest_multiple(low, high):
    candidates = np.arange(2520, 1e9, 2520)
    listNums = np.array(np.arange(low, high+1, 1))
    sums = [(sum(objs%listNums) + (objs%2)) for objs in candidates]
    return candidates[sums.index(0)]

smallest_multiple(1, 20)
```

# Number 6: Sum Square Difference

The sum of the squares of the first ten natural numbers is,
$$ 1^2 + 2^2 + ... + 10^2 = 385.$$

The square of the sum of the first ten natural numbers is,
$$ (1 + 2 + ... + 10)^2 = 55^2 = 3025.$$

Hence the difference between the sum of the squares of the first ten natural numbers and the square of the sum is $3025-385=2640$.

Find the difference between the sum of the squares of the first one hundred natural numbers and the square of the sum.

```python
import numpy as np

def sum_square_difference(naturalNums):
    listNums = np.arange(1,naturalNums+1,1)
    sumSquares = sum(listNums*listNums)
    squareSum = sum(listNums)**2
    return (squareSum - sumSquares)

sum_square_difference(100)
```

# Number 7: 10 001st Prime
By listing the first six prime numbers: $2,3,5,7,11,$ and $13$, we can see that the $6$th prime is $13$.

What is the $10001$st prime number?

```python
import numpy as np
from math import prod

# function to check whether a number is prime but it is so slow
def isPrime(num):
    divisors = np.arange(2, num**.5, 1)
    removedNum = divisors[divisors != num]
    has_zero = 0 in [num%d for d in removedNum]
    if not has_zero:
        return True
    else:
        return False

# slowly climb up via counter reaching the 10 001st prime
order = 10001
counter = 0
number = 2
while counter < (order):
    if isPrime(number):
        counter += 1
        number += 1
    else:
        number += 1

# this is very primitive implementation
print(number - 1)
```

# Problem 9: Special Pythagorean Triplet

<p>A Pythagorean triplet is a set of three natural numbers, $a \lt b \lt c$, for which,
$$a^2 + b^2 = c^2.$$</p>
<p>For example, $3^2 + 4^2 = 9 + 16 = 25 = 5^2$.</p>
<p>There exists exactly one Pythagorean triplet for which $a + b + c = 1000$.<br>Find the product $abc$.</p>

```python
import numpy as np

sumTriplet = 1000

row = np.array(np.arange(1, sumTriplet, 1))

matrix = [row, row, row]

# given the sum value condition, identify the total number of possible combinations of a, b, and c
def combiCalculator(sumTriplet):
    dimTotal = 0
    for i in range(1, sumTriplet):
        dimTotal += sum(range(1,i))
    return dimTotal

dim = combiCalculator(sumTriplet)
newMat = np.zeros((dim, 5)) # number of rows represent the combinations and cols represent the 3 values and 2 slots for the conditions

i = 0
for a in range(1, sumTriplet+1):
    for b in range(a+1, sumTriplet+1):
        for c in range(b+1, sumTriplet+1):
            newMat[i][0] = a
            newMat[i][1] = b
            newMat[i][2] = c
            newMat[i][3] = (a**2 + b**2 == c**2)*1.0
            newMat[i][4] = (a+b+c == sumTriplet)*1.0
            i += 1
            
tripletInd = np.nonzero(np.prod(newMat, axis=1))
tripletInd = tripletInd[0][0]
print(fr'The Pythagorean triplet for which a+b+c={sumTriplet} are...')
print(fr'a={newMat[tripletInd][0]}, b={newMat[tripletInd][1]}, c={newMat[tripletInd][2]}.')
print(fr'Their product is abc={newMat[tripletInd][0]*newMat[tripletInd][1]*newMat[tripletInd][2]}')
```

# Problem 10: Summation of Primes
<p>The sum of the primes below $10$ is $2 + 3 + 5 + 7 = 17$.</p>
<p>Find the sum of all the primes below two million.</p>

```python
import numpy as np

def sumPrimes(limit):
    
    """ Calculate the sum of all primes below a certain value. This is the Sieve of Erasthothenes method.
    Input: the limit or number for which we will calculate the sum of all primes below it.
    Output: sum """

    limit = int(limit) # convert limit to integer

    if limit < 0:
        raise ValueError('Input cannot be negative.')

    # create a list of boolean array
    isPrime = np.ones(limit, dtype=bool)
    isPrime[0:2] = False # set 0 and 1 to be not primes

    # run through from n=2 to sqrt(limit)
    # if n is True (or prime), make all multiples of n (starting from n^2) as False
    # after setting all multiples to False, proceed to next n
    for index in range(2, int(limit**.5)+1, 1): 
        if isPrime[index]:
            isPrime[index*index :: index] = False

    # the resulting array is a boolean array representing indices of prime numbers
    # convert the boolean array to corresponding countables
    return np.sum(isPrime*np.arange(0, limit, 1))

sumPrimes(2e6)
```

```python
def calcProductAdjacents(matrix, ajdNums=4):
    """ Given a matrix, calculate the greatest product of four adjacent numbers in the same direction.
    Input: matrix and number of adjacent numbers to consider
    Output: greatest product
 """
    rows, cols = matrix.shape
    horizProd, vertiProd, diagProd, antidiagProd = np.copy(matrix), np.copy(matrix), np.copy(matrix), np.copy(matrix)
    
    for i in range(adjNums-1):
        # shift horizontal
        horizProd *= np.roll(matrix0, shift=-i-1, axis=1)

        # shift vertical
        vertiProd *= np.roll(matrix0, shift=-i-1, axis=0)

        # shift diagonal
        matrix = np.roll(matrix0, shift=-i-1, axis=1)
        matrix = np.roll(matrix, shift=-i-1, axis=0)
        diagProd = diagProd * matrix

        # shift antidiagonal
        matrix = np.roll(matrix0, -(i+1), 0)
        matrix = np.roll(matrix, i+1, 1)
        antidiagProd = antidiagProd * matrix

    arrs = [horizProd, vertiProd, diagProd, antidiagProd]
    return np.max([np.max(arr) for arr in arrs])

calcProductAdjacents(matrix0)
```

# Problem 12: Highly Divisible Triangular Number

<p>The sequence of triangle numbers is generated by adding the natural numbers. So the $7$<sup>th</sup> triangle number would be $1 + 2 + 3 + 4 + 5 + 6 + 7 = 28$. The first ten terms would be:
$$1, 3, 6, 10, 15, 21, 28, 36, 45, 55, \dots$$</p>
<p>Let us list the factors of the first seven triangle numbers:</p>
$$\begin{align}
\mathbf 1 & \colon 1\\
\mathbf 3 & \colon 1,3\\
\mathbf 6 & \colon 1,2,3,6\\
\mathbf{10} & \colon 1,2,5,10\\
\mathbf{15} & \colon 1,3,5,15\\
\mathbf{21} & \colon 1,3,7,21\\
\mathbf{28} & \colon 1,2,4,7,14,28
\end{align}$$
<p>We can see that $28$ is the first triangle number to have over five divisors.</p>
<p>What is the value of the first triangle number to have over five hundred divisors?</p>


```python
import numpy as np

# to generate a sequence of triangle numbers given length of sequence and starting from 1
# started here since problem showed sequence 1 to 28
def triangleNumbersGenerator(length):
    """
    input: N index
    output: an array of a sequence of N triangular numbers from 1
    """
    length = int(length)
    listNums = np.arange(1, length+1, 1)
    listTri = []
    for num in listNums:
        sumVal = 0
        for i in range(num + 1):
            sumVal += i
        listTri.append(sumVal)
    return np.array(listTri)

# created a function to get factors for each number from a given a list
def getFactors(numList):
    arrayFactors = np.empty(len(numList), dtype=object)
    for i, num in enumerate(numList):
        divisors = np.arange(1, num+1, 1)
        arrayFactors[i] = (divisors[num%divisors == 0])
    return arrayFactors

# initially wanted to generate an array of triangle numbers
# then check the factors
# but the fact that the divisors in question is 500... it's a scaling problem now
triList = triangleNumbersGenerator(100)
triList = triList[:]
listFactors = getFactors(triList)
countFactors = np.array([len(factorTuple) for factorTuple in listFactors])
howManyDivisors = 5
triList[np.where(countFactors > howManyDivisors)[0][0]]
```

```python
# with the help of gemini, this is a more optimized technique


def count_divisors(n: int) -> int:
    """Counts total divisors of a number n"""
    count = 0
    d = 1
    while d * d <= n:
        if n % d == 0:
            count += 1
            if d * d != n:  # Avoid double-counting perfect square roots
                count += 1
        d += 1
    return count

def first_triangle_with_divisors(min_divisors: int):
    """Finds the first triangle number with strictly more than min_divisors."""
    n = 1
    while True:
        # Since n and n+1 are coprime, factors(T_n) = factors(n/2) * factors(n+1)
        if n % 2 == 0:
            divisors = count_divisors(n // 2) * count_divisors(n + 1)
        else:
            divisors = count_divisors(n) * count_divisors((n + 1) // 2)
            
        if divisors > min_divisors:
            triangle_num = n * (n + 1) // 2
            return n, triangle_num, divisors
        n += 1

how_many_divisors = 500
index, tri_val, num_factors = first_triangle_with_divisors(how_many_divisors)

print(f'Index={index}, triangle number {tri_val} with {num_factors} factors')
```

# Problem 14: Longest Collatz sequence

<p>The following iterative sequence is defined for the set of positive integers:</p>
<ul style="list-style-type:none;">
<li>$n \to n/2$ ($n$ is even)</li>
<li>$n \to 3n + 1$ ($n$ is odd)</li></ul>
<p>Using the rule above and starting with $13$, we generate the following sequence:
$$13 \to 40 \to 20 \to 10 \to 5 \to 16 \to 8 \to 4 \to 2 \to 1.$$</p>
<p>It can be seen that this sequence (starting at $13$ and finishing at $1$) contains $10$ terms. Although it has not been proved yet (Collatz Problem), it is thought that all starting numbers finish at $1$.</p>
<p>Which starting number, under one million, produces the longest chain?</p>
<p class="note"><b>NOTE:</b> Once the chain starts the terms are allowed to go above one million.</p>


```python
import numpy as np

def countCollatzChain(n):
    """ count the number of iterations required for a given number n to reach 1.0 using the collatz rules
    """
    count = 0
    while n != 1:
        if n%2:
            n = 3*n + 1
            count += 1
        else:
            n /= 2
            count += 1
    return count + 1


limit = 1e6

listNums = np.arange(1, 1e6+1, 1)

chain = 0
number = 0
for num in listNums:
    chainNew = countCollatzChain(num)
    if chainNew > chain:
        number, chain = num, chainNew
    else:
        continue

print(number)
```
