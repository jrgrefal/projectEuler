# Number 1

```python
maxNum = 999
toSum = [] 
for i in range(maxNum + 1):
    if (i%3 and i%5) == 0:
        toSum.append(i)
np.sum(toSum)
```

```python
maxNum = 1000
listOfNums = np.arange(1,maxNum,1)

indices = listOfNums%3*listOfNums%5
newList = listOfNums[indices == 0]

np.sum(newList)
```

```python
start1 = 1
start2 = 2
terms = 10000 - 2
nextNum = start1 + start2

listOfEvenParticulars = [1, 2]

while nextNum < 4e6:
    listOfEvenParticulars.append(nextNum)
    start1, start2 = start2, nextNum 
    nextNum = start1 + start2

indices = np.array(listOfEvenParticulars)%2
np.sum(np.array(listOfEvenParticulars)[indices == 0])
```
